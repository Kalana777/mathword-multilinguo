```python
import pandas as pd
```

```python
gemma_df = pd.read_excel("outputs/gemma-multilingual-0s-responses_11.xlsx")
qwen_df = pd.read_excel("outputs/qwen-multilingual-0s-responses.xlsx")
aya_df = pd.read_excel("outputs/aya-multilingual-0s-responses.xlsx")
krut_df = pd.read_excel("outputs/krutrim-multilingual-0s-responses.xlsx")
```

```python
sinhala_gemma = gemma_df[gemma_df["language"]=="Sinhala"].copy().reset_index(drop=True)
tamil_gemma = gemma_df[gemma_df["language"]=="Tamil"].copy().reset_index(drop=True)
albanian_gemma = gemma_df[gemma_df["language"]=="Albanian"].copy().reset_index(drop=True)
hindi_gemma = gemma_df[gemma_df["language"]=="Hindi"].copy().reset_index(drop=True)
punjabi_gemma = gemma_df[gemma_df["language"]=="Punjabi"].copy().reset_index(drop=True)
marathi_gemma = gemma_df[gemma_df["language"]=="Marathi"].copy().reset_index(drop=True)
odia_gemma = gemma_df[gemma_df["language"]=="Odia"].copy().reset_index(drop=True)
```

```python
sinhala_qwen = qwen_df[qwen_df["language"]=="Sinhala"].copy().reset_index(drop=True)
tamil_qwen = qwen_df[qwen_df["language"]=="Tamil"].copy().reset_index(drop=True)
albanian_qwen = qwen_df[qwen_df["language"]=="Albanian"].copy().reset_index(drop=True)
hindi_qwen = qwen_df[qwen_df["language"]=="Hindi"].copy().reset_index(drop=True)
punjabi_qwen = qwen_df[qwen_df["language"]=="Punjabi"].copy().reset_index(drop=True)
marathi_qwen = qwen_df[qwen_df["language"]=="Marathi"].copy().reset_index(drop=True)
odia_qwen = qwen_df[qwen_df["language"]=="Odia"].copy().reset_index(drop=True)
```

```python
sinhala_aya = aya_df[aya_df["language"]=="Sinhala"].copy().reset_index(drop=True)
tamil_aya = aya_df[aya_df["language"]=="Tamil"].copy().reset_index(drop=True)
albanian_aya = aya_df[aya_df["language"]=="Albanian"].copy().reset_index(drop=True)
hindi_aya = aya_df[aya_df["language"]=="Hindi"].copy().reset_index(drop=True)
punjabi_aya = aya_df[aya_df["language"]=="Punjabi"].copy().reset_index(drop=True)
marathi_aya = aya_df[aya_df["language"]=="Marathi"].copy().reset_index(drop=True)
odia_aya = aya_df[aya_df["language"]=="Odia"].copy().reset_index(drop=True)
```

```python
sinhala_krut = krut_df[krut_df["language"]=="Sinhala"].copy().reset_index(drop=True)
tamil_krut = krut_df[krut_df["language"]=="Tamil"].copy().reset_index(drop=True)
albanian_krut = krut_df[krut_df["language"]=="Albanian"].copy().reset_index(drop=True)
hindi_krut = krut_df[krut_df["language"]=="Hindi"].copy().reset_index(drop=True)
punjabi_krut = krut_df[krut_df["language"]=="Punjabi"].copy().reset_index(drop=True)
marathi_krut = krut_df[krut_df["language"]=="Marathi"].copy().reset_index(drop=True)
odia_krut = krut_df[krut_df["language"]=="Odia"].copy().reset_index(drop=True)
```

```python
with pd.ExcelWriter("outputs/processed/sinhala_zero.xlsx", engine="openpyxl") as writer:
    sinhala_gemma.to_excel(writer, sheet_name="gemma", index=False)
    sinhala_qwen.to_excel(writer, sheet_name="qwen", index=False)
    sinhala_aya.to_excel(writer, sheet_name="aya", index=False)
    sinhala_krut.to_excel(writer, sheet_name="krutrim", index=False)
```
