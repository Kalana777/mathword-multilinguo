```python
import pandas as pd
from string import Template
from pathlib import Path

DATA_FILE = Path("data/odia.xlsx")
OUTPUT_FILE = Path("data/incontext_odia.csv")

TEMPLATE_FILE_ZERO = Path("qwen_zeroshot_template.txt")
TEMPLATE_FILE_ONESHOT = Path("qwen_1s_template.txt")
```

```python
df = pd.read_excel(DATA_FILE)
zero_template_text = TEMPLATE_FILE_ZERO.read_text(encoding="utf-8")
one_template_text = TEMPLATE_FILE_ONESHOT.read_text(encoding="utf-8")

tmpl0 = Template(zero_template_text)
tmpl1 = Template(one_template_text)
```

```python
df.head()
```

```python
filled_rows = []

for idx, row in df.iterrows():

    row_dict = row.to_dict()
    mapping = {}
    for k, v in row_dict.items():
        if pd.isna(v):
            mapping[k] = ""
        else:
            mapping[k] = str(v)

    try:
        out_0 = tmpl0.substitute(mapping)
        out_1 = tmpl1.substitute(mapping)
    except KeyError as e:
        raise KeyError(f"Missing placeholder value for: {e.args[0]}") from None

    row_dict["zeroshot_prompt"] = out_0
    row_dict["oneshot_prompt"] = out_1
    filled_rows.append(row_dict)

df_out = pd.DataFrame(filled_rows)
```
